# 00 - 环境检查

## 目的
确认 Python 环境和所有依赖库已正确安装。

## 输入
无

## 输出
- 各库版本信息
- 环境是否满足要求

## 步骤
1. 检查 Python 版本
2. 检查核心库是否安装
3. 检查数据目录是否存在

```python
import sys
print(f"Python version: {sys.version}")
```

```python
# 检查核心库
libs = ['pandas', 'numpy', 'matplotlib', 'seaborn', 'sklearn', 'xgboost', 'lightgbm']
for lib in libs:
    try:
        m = __import__(lib)
        print(f"{lib}: {m.__version__}")
    except ImportError:
        print(f"{lib}: NOT INSTALLED")
```

```python
# 检查数据目录
from pathlib import Path
data_dir = Path.cwd().parent / "data"
print(f"data/raw exists: {(data_dir / 'raw').exists()}")
print(f"data/processed exists: {(data_dir / 'processed').exists()}")
```
