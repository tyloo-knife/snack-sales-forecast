# 05 - 特征工程

## 目的
构造对预测有帮助的特征变量。

## 输入
- `data/processed/` 下的清洗后数据

## 输出
- 带有新特征的数据集
- 特征重要性分析

## 步骤
1. 时间特征（星期、月份、是否周末、是否节假日）
2. 滞后特征（前 1/7/14 天销量）
3. 滚动统计（7 天/14 天均值、标准差）
4. 编码分类变量
5. 特征选择

## 注意事项
- 特征构造时不能使用未来信息（data leakage）
- 滞后特征的窗口不能跨越训练/验证集边界
